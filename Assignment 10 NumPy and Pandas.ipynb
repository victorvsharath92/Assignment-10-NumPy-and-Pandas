```python
import pandas as pd
import numpy as np 
```

```python
# NUMPY ARRAY OPERATIONS
```

```python
# 1. Create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
# 2. Inspection and Properties
print("--- 1D Array Inspection ---")
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# 3. Array Operations
temps_w1_fahrenheit = (temperatures_w1 * 9 / 5) + 32
max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print("\n--- 1D Array Operations ---")
print("Fahrenheit:", temps_w1_fahrenheit)
print(f"Max: {max_temp}, Min: {min_temp}, Mean: {mean_temp:.2f}")
```

```python
# 4. Array Slicing and Indexing
first_three_days = temperatures_w1[:3]
weekend_days = temperatures_w1[-2:]
middle_three_days = temperatures_w1[2:5]

print("\n--- 1D Slicing ---")
print("First three days:", first_three_days)
print("Weekend:", weekend_days)
print("Middle three days:", middle_three_days)
```

```python
# 5. Create a 2D Array
temperatures = np.array(
    [
        [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
        [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1],
    ]
)
```

```python
# 6. Inspect and Slice the 2D Array
print("\n--- 2D Array Inspection ---")
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)

week_1 = temperatures[0]
week_2 = temperatures[1]
weekends_both_weeks = temperatures[:, -2:]

print("\n--- 2D Slicing ---")
print("Week 1:", week_1)
print("Week 2:", week_2)
print("Weekends for both weeks:\n", weekends_both_weeks)
```

```python
# 1. Creating Pandas Series

marks_data = [95, 92, 89, 85, 80]
indices = ["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]
marks = pd.Series(data=marks_data, index=indices)

print("\n--- Pandas Series ---")
print(marks)
```

```python
# 2. Indexing and Slicing

rank1_mark = marks.iloc[0]
top3_marks = marks.loc["Rank1":"Rank3"]
rank3_mark = marks.iloc[2]
above_90 = marks[marks > 90]

print("\n--- Series Indexing & Slicing ---")
print("1st rank mark (integer position):", rank1_mark)
print("Top 3 ranks (loc):\n", top3_marks)
print("3rd rank mark (iloc):", rank3_mark)
print("Marks > 90:\n", above_90)
```

```python
# 3. Manipulating Series
marks.loc["Rank1"] = 100
marks = marks.drop("Rank5")  # Drop last rank
cgpa = marks / 10

print("\n--- Series Manipulation ---")
print("Updated Marks:\n", marks)
print("CGPA:\n", cgpa)
```

```python
# 1. Creating Pandas DataFrame
data = {
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": [
        "Electronics",
        "Clothing",
        "Electronics",
        "Furniture",
        "Clothing",
        "Electronics",
        "Furniture",
        "Clothing",
        "Furniture",
        "Electronics",
    ],
    "Region": [
        "North",
        "South",
        "North",
        "East",
        "West",
        "North",
        "East",
        "West",
        "South",
        "North",
    ],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400],
}

transactions = pd.DataFrame(data)
```

```python
# 2. Data Exploration
print("\n--- DataFrame Exploration ---")
print("Head:\n", transactions.head())
print("\nTail:\n", transactions.tail())
print("\nShape:", transactions.shape)
print("Columns:", transactions.columns.tolist())
print("\nData Types:\n", transactions.dtypes)
```

```python
# Display specific columns
print("\nSelected Columns:\n", transactions[["ProductCategory", "Amount"]])
```

```python
# Retrieve the last 3 columns using iloc
print("\nLast 3 Columns:\n", transactions.iloc[:, -3:])
```

```python
# Filter rows: Region == 'North' and Amount > 200
filtered_transactions = transactions[
    (transactions["Region"] == "North") & (transactions["Amount"] > 200)
]
print("\nFiltered Rows (North & Amount > 200):\n", filtered_transactions)
```

```python
# Value counts for 'ProductCategory'
print(
    "\nProduct Category Counts:\n",
    transactions["ProductCategory"].value_counts(),
)
```

```python
# Unique values in 'Region'
print("\nUnique Regions:", transactions["Region"].unique())
```

```python
# Group by 'Region' and calculate mean 'Amount'
mean_amount_by_region = transactions.groupby("Region")["Amount"].mean()
print("\nMean Amount by Region:\n", mean_amount_by_region)
```

```python
# 3. Manipulating the DataFrame
# Modify Amount for TransactionID 102
transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165
```

```python
# Add 'Discount' column (10% of Amount)
transactions["Discount"] = transactions["Amount"] * 0.10
```

```python
# Remove the row where TransactionID is 109
transactions = transactions[transactions["TransactionID"] != 109].reset_index(
    drop=True
)
```

```python
# Delete 'Discount' column
transactions.drop(columns=["Discount"], inplace=True)
```

```python
print("\n--- Final Manipulated DataFrame ---")
print(transactions)
```
